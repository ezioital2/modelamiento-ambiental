# Ejercicio 3

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

D0 = 1.58
L0 = 8.0
kd = 0.25
kr = 0.40
ODs = 9.08

t = np.array([0, 1, 2, 4, 6, 8, 10])
L = L0 * np.exp(-kd * t)
D = kd * L0 / (kr - kd) * (np.exp(-kd * t) - np.exp(-kr * t)) + D0 * np.exp(-kr * t)
OD = ODs - D

tabla = pd.DataFrame({"t": t, "L(t)": L, "D(t)": D, "OD(t)": OD}).round(2)
tabla
```

```python
plt.plot(t, D, "o-", label="D(t)")
plt.plot(t, OD, "o-", label="OD(t)")
plt.xlabel("t (d)")
plt.ylabel("mg/L")
plt.legend()
plt.grid()
plt.show()
```

```python
tc = np.log(kr / kd * (1 - D0 * (kr - kd) / (kd * L0))) / (kr - kd)
j = np.searchsorted(t, tc)
print("tc =", round(tc, 2), "d")
print("El mínimo de OD aparece entre t =", t[j-1], "y t =", t[j], "d")
```

# Ejercicio 5

```python
v = 0.50

tc_s = tc * 86400
xc_m = v * tc_s
xc_km = xc_m / 1000

pd.DataFrame({"Resultado": ["tc", "xc", "xc"],
              "Valor": [round(tc_s, 0), round(xc_m, 0), round(xc_km, 2)],
              "Unidad": ["s", "m", "km"]})
```

**Significado ambiental de la distancia crítica:** es el punto del río, aguas abajo de la descarga, donde el oxígeno disuelto llega a su valor mínimo (máximo déficit). Es la zona más afectada para la vida acuática y donde conviene monitorear y verificar que el OD no baje del límite permitido; después de ese punto el río empieza a recuperarse por reaireación.
